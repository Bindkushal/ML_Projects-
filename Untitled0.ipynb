```python
import numpy as np
import matplotlib.pyplot as plt

# Generate x values from 0 to 2*pi
x = np.linspace(0, 2 * np.pi, 100)

# Calculate sin(x)
y = np.sin(x)

# Create the plot
plt.figure(figsize=(8, 4))
plt.plot(x, y)
plt.title('Graph of sin(x)')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.grid(True)
plt.show()
```

### 1. Import Libraries
We'll start by importing the necessary libraries: `tensorflow` for the neural network, `numpy` for data generation, and `matplotlib` for visualization.

```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt

print(f"TensorFlow Version: {tf.__version__}")
```

### 2. Generate Synthetic Data
For this example, we'll create two clusters of 2D data points, representing two different classes. We'll add some noise to make it more realistic.

```python
# Number of samples per class
n_samples = 100

# Class 0: centered around (0, 0)
X_0 = np.random.randn(n_samples, 2) * 0.5 + np.array([0, 0])
y_0 = np.zeros(n_samples)

# Class 1: centered around (2, 2)
X_1 = np.random.randn(n_samples, 2) * 0.5 + np.array([2, 2])
y_1 = np.ones(n_samples)

# Combine data
X = np.vstack((X_0, X_1))
y = np.hstack((y_0, y_1))

# Shuffle the data
indices = np.arange(len(y))
np.random.shuffle(indices)
X = X[indices]
y = y[indices]

# Plot the generated data
plt.figure(figsize=(6, 5))
plt.scatter(X[y == 0][:, 0], X[y == 0][:, 1], label='Class 0', alpha=0.7)
plt.scatter(X[y == 1][:, 0], X[y == 1][:, 1], label='Class 1', alpha=0.7)
plt.title('Synthetic 2D Data for Classification')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```

### 3. Build the Neural Network Model
We'll create a simple Sequential model with two `Dense` (fully connected) layers. The first layer takes the 2 features as input, and the output layer has 1 neuron with a sigmoid activation for binary classification.

```python
model = keras.Sequential([
    keras.layers.Dense(units=4, activation='relu', input_shape=(2,)), # Hidden layer with 4 neurons and ReLU activation
    keras.layers.Dense(units=1, activation='sigmoid') # Output layer with 1 neuron and Sigmoid activation for binary classification
])

model.summary()
```

### 4. Compile the Model
Compiling the model involves specifying the optimizer, loss function, and metrics. For binary classification, we typically use `binary_crossentropy` as the loss function and `adam` as the optimizer.

```python
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])
```

### 5. Train the Model
Now, let's train the model using our synthetic data. We'll train for a few epochs.

```python
history = model.fit(X, y, epochs=50, verbose=0) # verbose=0 to suppress output for each epoch

# Plot training history (loss and accuracy)
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['loss'])
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'])
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.grid(True)

plt.tight_layout()
plt.show()
print(f"Final training accuracy: {history.history['accuracy'][-1]:.4f}")
```

### 6. Make Predictions
After training, we can use the model to make predictions on new data. Let's predict on the training data itself to see how well it learned.

```python
predictions = model.predict(X)
predicted_classes = (predictions > 0.5).astype(int)

# Plot the decision boundary (for visualization purposes)
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1),
                       np.arange(y_min, y_max, 0.1))

Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = (Z > 0.5).astype(int).reshape(xx.shape)

plt.figure(figsize=(7, 6))
plt.contourf(xx, yy, Z, alpha=0.4)
plt.scatter(X[y == 0][:, 0], X[y == 0][:, 1], label='Class 0', alpha=0.7, edgecolors='k')
plt.scatter(X[y == 1][:, 0], X[y == 1][:, 1], label='Class 1', alpha=0.7, edgecolors='k')
plt.title('Neural Network Decision Boundary')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```
